```python
def extract_pages(pdf_path):
    # open the file
    # build the list of (page_number, text)
    # return the list

    doc = pymupdf.open(pdf_path)

    pages = []
    for page_number, page in enumerate(doc, start = 1):
        #print(page_number)
        blocks = page.get_text("blocks", sort=True)
        text = "\n\n".join([block[4].strip() for block in blocks])
        pages.append((page_number, text))
    return pages
```

```python
def search_corpus(keyword):
    for game, pages in all_pages.items():
        for page_number, text in pages:
            # if keyword is in text:
            #     find its position
            #     print game, page_number, and ~150 characters around it
            if keyword in text:
                position = text.find(keyword)
                start = max(0, position - 75)
                end = min(len(text), position + len(keyword) + 75)
                snippet = text[start:end]
                print(f"Game: {game}, Page: {page_number}")
                print(snippet)
                print("\n")
```

## Cleaning plan

| # | Problem | Example | Fix type | Why |
|---|---|---|---|---|
| 1 | Currency letters | `a200`, `b50`, `A6` | clean | Info is in the text with the wrong symbol; a pattern can rewrite it as `M200` |
| 2 | Control characters | `\x07`, `\x03` | clean | Invisible junk with no meaning; delete |
| 3 | Print specs & product codes | `700 x 161 mm…`, `C1009`, `0622C1009780` | clean | Printer info, never answers a rule question; remove known lines |
| 4 | Repeated copyright & slogans | `© 2025 CATAN GmbH` ×4 | clean | Duplicates crowd real rules out of top-k; remove |
| 5 | Manufacturer address & trademark text | `Hasbro SA, Rue Emile-Boéchat…` | clean | Legal text, not rules; remove |
| 6 | Jumbled reading order | page 9 began with card text | parser | Fixed by sorted blocks |
| 7 | Side boxes pushed into sentences | `stay BUILDING COST: hidden` | parser | Fixed by blocks: each box is its own block |
| 8 | Spaced-out letters | `K L A U S` | keep | Only on the cover; no rule content affected |
| 9 | Card-picture text duplicates body text | `Take any 2 resource cards… INVENTION` | keep (measure) | Hard to remove safely; check later if duplicates hurt retrieval |
| 10 | Card text before its title | description, then `INVENTION` | keep (measure) | Comes from the card pictures (#9); body paragraphs are in the right order |
| 11 | Missing icons | `Rent with` + blank + `A300`; empty `BUILDING COST:` | can't fix | Info is in pictures, not text; expect eval rows 6 and 12 to fail; later: captions or a vision model |
| 12 | Mid-sentence line breaks | `not \neach other` | clean | Join lines inside a block into one paragraph |
| 13 | Zero-width chars & look-alike hyphens | `●​`, `face‑up` | clean | Normalize so user queries match the text exactly |
| 14 | Rummy suit symbols missing | `-9` gaps | can't fix | Lost in the Google Docs copy; re-export from the web page as a PDF |
| 15 | Intro & history paragraphs | "Pochen", riverboats | keep | Not wrong, just low-value; measure whether they crowd out rules |

```python
REPLACEMENTS = {
    "\u2011": "-",   # non-breaking hyphen → normal hyphen
    "\u200b": "",    # zero-width space → delete
    "\x07": "",      # control character → delete
    "\x03": "",      # control character → delete
    "\u2019": "'",   # curly apostrophe → straight (don’t → don't)
}

def normalize_characters(text):
    for old, new in REPLACEMENTS.items():
        text = text.replace(old, new)
    return text
```

```python
def fix_currency(text, game):
    if game != "monopoly":
        return text
    return re.sub(r"\b[aAb](\d+)", r"M\1", text)
```

```python
JUNK_MARKERS = [
    "4C2S",                       # print spec: "700 x 161 mm, 4C2S Folded..."
    "©",                          # copyright lines (Catan and Monopoly)
    "Hasbro SA",                  # manufacturer address
    "Represented by",
    "P.O. Box",
    "consumer_affairs",
    "Fast-Dealing Property Trading Game",  # repeated slogan
    "are trademarks of Hasbro",
]
```

```python
def remove_junk_lines(text):
    blocks = text.split("\n\n")
    kept = []
    for block in blocks:
        if any(marker in block for marker in JUNK_MARKERS):
            continue
        kept.append(block)
    return "\n\n".join(kept)
```

```python
original = all_pages["monopoly"][0][1]
cleaned = remove_junk_lines(original)
removed = [l for l in original.split("\n") if l not in cleaned.split("\n")]
for l in removed:
    print("REMOVED:", l)
```

```python
def join_broken_lines(text):
    blocks = text.split("\n\n")          # separate the paragraphs
    kept = []
    for block in blocks:
        block = block.replace("\n", " ")       # 1. join the broken lines
        block = re.sub(r" +", " ", block)      # 2. squeeze double spaces
        block = block.strip()                  # 3. trim the ends
        kept.append(block)                     # 4. keep it
    return "\n\n".join(kept)             # paragraphs back, separated by blank lines
```

```python
print(join_broken_lines(all_pages["blackjack"][0][1])[:500])
```

```python
def clean_text(text, game):
    text = normalize_characters(text)   # rows 2 + 13
    text = fix_currency(text, game)     # row 1
    text = remove_junk_lines(text)      # rows 3, 4, 5
    text = join_broken_lines(text)      # row 12
    return text
```

```python
cleaned_pages = {}

for game, pages in all_pages.items():
    cleaned_pages[game] = []
    for page_number, text in pages:
        cleaned_text = clean_text(text, game)
        cleaned_pages[game].append((page_number, cleaned_text))


print(len(cleaned_pages))
print(cleaned_pages["monopoly"][1][1][:500])
```
