## Monopoly findings

1. page is divided into boxes like tuples and tuples then into words and characters. for example pages[0][1][:500] means ipages is a list of tuples, and in each tuple [0] is the page number and [1] is the text. So [1] means “the text,” not “the second text.
2. There are some problems here that needs a cleaning step. 
    a. Junk text(print specs, \x07, copyright lines)
    b. Currency letters (a200, b50)
3. Also, there are some which cannot be done with cleaning are:
    
    a. Missing icons (rent rows, building costs)



```python
def extract_pages(pdf_path):
    # open the file
    # build the list of (page_number, text)
    # return the list

    doc = pymupdf.open(pdf_path)

    pages = []
    for page_number, page in enumerate(doc, start = 1):
        #print(page_number)
        blocks = page.get_text("blocks", sort=True)
        text = "\n\n".join([block[4].strip() for block in blocks])
        pages.append((page_number, text))
    return pages
```

```python
catan_pages = extract_pages("../data/raw/catan.pdf")
print(len(catan_pages))
print(type(catan_pages[8][1][:500]))

print(catan_pages[8][1][:500])
```

## Catan findings

spaced letters (K L A U S), card text before its title, the MONOPOLY card colliding with the Monopoly game, the repeated copyright line, the empty BUILDING COST, the duplicate card-picture text

“I chose sorted blocks because you just used sort=True: get_text("blocks", sort=True). What broke sentences before was sorting line by line. With blocks, sorting moves whole paragraphs, so sentences stay intact. So the right conclusion isn’t “no sort.” It’s:
Use blocks, sorted by position. Paragraphs stay whole, and the order matches how a person reads the page.




```python
pdf_paths = {
    "catan": "../data/raw/catan.pdf",
    "monopoly": "../data/raw/monopoly.pdf",
    "blackjack": "../data/raw/blackjack.pdf",
    "poker": "../data/raw/poker.pdf",
    "rummy": "../data/raw/rummy.pdf",
    "texas_holdem": "../data/raw/texas_holdem.pdf",
}

all_pages = {}
for game, pdf_path in pdf_paths.items():
    all_pages[game] = extract_pages(pdf_path)
    print(f"{game}: {len(all_pages[game])} pages")
```
